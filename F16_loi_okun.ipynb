#Étude de la loi d'Okun
##1. Présentation du projet

Problématique :
Dans quelle mesure la croissance du PIB réel permet-elle d'expliquer la variation du taux de chômage aux États-Unis ?

Type d'apprentissage :
 Apprentissage supervisé — Régression

Algorithme utilisé :
Régression linéaire simple (Linear Regression)

Source des données :
 FRED (Federal Reserve Economic Data)

 **cible à prédire (Y) :**

La variation trimestrielle du taux de chômage.

**Variable explicative (X) :**

Le taux de croissance trimestriel du PIB réel.

**Tâche** :

L'objectif est d'utiliser la régression linéaire simple (Linear Regression) pour étudier la relation entre la croissance du PIB réel et la variation du taux de chômage, puis essayer de prédire cette variation à partir de la croissance économique.

**Métrique de réussite :**

Pour évaluer notre modèle, nous allons utiliser le coefficient de détermination (R^2). Nous fixons comme objectif d'obtenir un (R^2) supérieur ou égal à 0,50, ce qui signifie que le modèle explique au moins 50 % des variations du taux de chômage.

## 2. Présentation des données

Pour réaliser cette étude, nous allons utiliser deux séries statistiques provenant de la base de données FRED (*Federal Reserve Economic Data*), qui met à disposition des données économiques et financières.

Les données concernent les États-Unis et portent sur les deux variables suivantes :

**1. Le produit intérieur brut réel (PIB réel)**

- **Code de la série :** GDPC1
- **Description :** Produit intérieur brut réel des États-Unis.
- **Fréquence :** Trimestrielle.
- **Rôle dans l'étude :** Cette variable permettra de mesurer la croissance économique.

**2. Le taux de chômage**

- **Code de la série :** UNRATE
- **Description :** Taux de chômage aux États-Unis.
- **Fréquence :** Mensuelle.
- **Rôle dans l'étude :** Cette variable permettra d'observer l'évolution du chômage.

**Préparation des données :**

Les deux séries n'ont pas la même fréquence. Le PIB réel est trimestriel, tandis que le taux de chômage est mensuel. Il faudra donc convertir les données mensuelles du chômage en données trimestrielles, en calculant la moyenne des trois mois de chaque trimestre.

Ensuite, nous calculerons le taux de croissance du PIB réel et la variation du taux de chômage afin d'étudier leur relation.
